# **Tugas**

```python
import random

size =int(input("Masukkan ukuran peta: "))

treasure_map = ["_" for _ in range(size)]

print(f"Kamu telah memulai Treasure Hunt ! Peta memiliki {size} lokasi (0 hingga {size-1}). ")
treasure_index = random.randint(0, size-1)
print(f"Treasure disembunyikan secara acak. Cobalah untuk menemukannya!")
while treasure_index < 0 or treasure_index >= size:
  print("Lokasi tidak ditemukan. Masukkan lokasi yang termasuk ke dalam peta.")
  treasure_index = int(input(f"Masukan lokasi treasure (0 hingga {size-1}): "))

treasure_map[treasure_index] = "X"

def play_treasure_hunt():
  print ("\nTreasure telah disembunyikan! Sekarang giliran pemain menebak")
  print ("Cobalah untuk menemukan Treasure dengan menebak lokasi yang benar.")
  while True:
    print ("\nPeta saat ini: ", ["_" for _ in range(size)])
    try:
      guess = int(input(f"Masukkan lokasi tebakanmu (0 hingga {size-1}): "))
    except ValueError:
      print("Masukkan lokasi yang valid.")
      continue

    if guess < 0 or guess >= size:
      print("lokasi tidak valid, pilih antara 0 hingga", size-1)
      continue

    if treasure_map[guess] == "X":
      print(" ")
      print("Yeay Berhasil! Treasure ditemukan di lokasi", guess)
      break
    else:
      print("Tidak ada Treasure di lokasi ini. Ayo coba lagi!")

play_treasure_hunt()
```
